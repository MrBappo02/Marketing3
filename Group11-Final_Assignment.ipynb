# Marketing 3 — CRISP-DM workshop

**Business context:** Local grocery loyalty campaigns

**Dataset:** group_11_marketing_3.csv


## 1. Business Understanding
Define the business problem, stakeholder/user, decision, prediction/decision moment, scope and measurable success criteria. Critically assess scenario assumptions. Do not invent company facts.

## 2. Data Understanding
Establish observation unit, key, target and variable meanings. Profile quality and distributions. Identify patterns, anomalies, limitations and possible leakage. Explain how evidence confirms, challenges or refines Phase 1.

## 3. Data Preparation
Make at least three substantive preparation decisions, including at least one feature inclusion/exclusion/engineering decision. For important choices show evidence → technique → before/after verification → interpretation → next decision. Explicitly prevent leakage.

## 4. Modelling
Create an appropriate baseline and at least one justified candidate model. Use a validation strategy and metrics that fit the business problem and data structure. Model sophistication is not rewarded by itself.

## 5. Evaluation
Return to Phase 1. Compare candidate and baseline, interpret errors in business terms, inspect relevant segments/subgroups, discuss limitations and decide: deploy, pilot, revise, collect more data, return to an earlier phase, or do not deploy.

## 6. Deployment / Next Iteration
Specify who uses the output, when it is produced, what action it supports, human oversight/override, monitoring, feedback/refresh trigger and a pause/stop condition. You design deployment; you do not need to build a production system.

**Marketing:** handle mixed units/categories and post-campaign leakage; evaluate under a stated contact-capacity assumption; distinguish conversion prediction from causal uplift.
